```python
import pandas as pd
import numpy as np
```

```python
f = pd.read_excel(r"C:\Users\vartika gautam\Downloads\SQL_Sales_Dataset_200_Rows.xlsx")
```

```python
# Display basic information about the dataset
print(f.info())

# Display the first 5 rows
print(f.head())
```

```python
# Check for missing values and duplicates
print(f.isnull().sum())
print("Duplicate rows:", f.duplicated().sum())

# Drop duplicate rows
f = f.drop_duplicates()

# Handle missing values (e.g., fill numerical columns with mean, categorical with mode)
f["total_price"] = f["total_price"].fillna(f["total_price"].mean())
f["category"] = f["category"].fillna("Unknown")

```

```python
# Group by 'category' and sum the 'revenue' column
revenue_by_category = f.groupby("category")["total_price"].sum().reset_index()

print(revenue_by_category)
```

```python
# Sort the DataFrame by 'category' in ascending order and 'revenue' in descending order
sorted_df = f.sort_values(by=["category", "total_price"], ascending=[True, False])

print(sorted_df.head(10))
```

```python
# Calculate the correlation matrix for all numerical columns
correlation_matrix = f.select_dtypes(include=["number"]).corr()

print(correlation_matrix)
```
